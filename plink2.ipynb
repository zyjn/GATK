```python
import os
import subprocess

print("🔄 正在清理旧版本并重装兼容版 (Standard 64-bit)...")

# 1. 删除旧文件 (防止干扰)
!rm -f plink2 plink2.zip *.zip

# 2. 下载兼容版 (Linux 64-bit Intel, 非 AVX2)
# 注意：这里使用的是 x86_64 标准版，虽然速度微慢一点点，但在所有机器上都能跑
!wget https://s3.amazonaws.com/plink2-assets/alpha6/plink2_linux_avx2_20260110.zip

# 3. 解压
!unzip -o plink2_linux_avx2_20260110.zip

# 4. 赋予执行权限
!chmod +x plink2

# 5. 验证
if os.path.exists("plink2"):
    abs_path = os.path.abspath("plink2")
    print(f"\n✅ 重装完成！兼容版绝对路径: {abs_path}")
    
    print("⬇️ 版本测试 (这次应该有输出了):")
    # 使用 check_output 强制捕获输出，如果有错会显示出来
    try:
        output = subprocess.check_output("./plink2 --version", shell=True, stderr=subprocess.STDOUT)
        print(output.decode("utf-8"))
    except subprocess.CalledProcessError as e:
        print(f"❌ 运行报错: {e.output.decode('utf-8')}")
else:
    print("❌ 下载或解压失败，请检查网络。")
```
